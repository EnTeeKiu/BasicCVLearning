## CV Array Conventions

Common image shapes:

- grayscale image: `[H, W]`
- color image in NumPy/PIL/OpenCV: `[H, W, C]`
- PyTorch image tensor: `[C, H, W]`
- batch of PyTorch images: `[B, C, H, W]`

Common dtypes:

- raw image from disk: `uint8`, usually range `0..255`
- model input: `float32`, usually range `0..1`, then optionally normalized
