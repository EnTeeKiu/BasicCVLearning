## CV Array Conventions

Common image shapes:

- grayscale image: `[H, W]`
- color image in NumPy/PIL/OpenCV: `[H, W, C]`
- PyTorch image tensor: `[C, H, W]`
- batch of PyTorch images: `[B, C, H, W]`

Common dtypes:

- raw image from disk: `uint8`, usually range `0..255`
- model input: `float32`, usually range `0..1`, then optionally normalized


```python
import random
import numpy as np

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
```

## Exercise 04-A: dtype and Normalization

Convert `[0, 127, 255]` from `uint8` to float values in `[0, 1]`.


```python
u8 = np.array([0, 127, 255], dtype=np.uint8)
out = u8.astype(np.float32) / 255.0
print(out)
print(out.dtype)
```

## Exercise 04-B: HWC to CHW

PyTorch expects channel-first images. Convert an image from `[H, W, C]` to `[C, H, W]`.


```python
image = np.arange(2 * 3 * 3).reshape(2, 3, 3)
chw = np.transpose(image, (2, 0, 1))
print("original:", image.shape)
print("converted:", chw.shape)
print("first channel:")
print(chw[0])
```

## Exercise 04-C: Batch Construction

A batch of two RGB images for PyTorch should be `[B, C, H, W]`.


```python
imgs = [
    np.zeros((4, 5, 3), dtype=np.uint8),
    np.full((4, 5, 3), 255, dtype=np.uint8),
]
chw_imgs = [np.transpose(img, (2, 0, 1)) for img in imgs]
batch = np.stack(chw_imgs, axis=0)
print(batch.shape)
print(batch.dtype)
```

## Exercise 04-D: Mean/Std Normalization

ImageNet-style normalization subtracts a per-channel mean and divides by a per-channel std. Broadcasting is your friend here.


```python
img = np.ones((2, 2, 3), dtype=np.float32)
mean = np.array([0.5, 1.0, 1.5], dtype=np.float32)
std = np.array([0.5, 0.5, 0.5], dtype=np.float32)
out = (img - mean) / std
print(out)
print(out.shape)
```

## Exercise 04-E: Channel Order

Simulate one RGB pixel and convert it to BGR. This is the tiny version of the PIL/OpenCV mismatch.


```python
rgb_pixel = np.array([[[255, 0, 0]]], dtype=np.uint8)
bgr_pixel = rgb_pixel[..., ::-1]
print("RGB:", rgb_pixel)
print("BGR:", bgr_pixel)
```
