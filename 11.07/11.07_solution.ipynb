Today is about making image preprocessing reliable: resize, crop, flip, color jitter, tensor conversion, and normalization. The main habit is separating stochastic training transforms from deterministic validation transforms.


## Core Ideas

Transforms convert raw images into model-ready tensors while controlling what variation the model sees.

- **Resize:** makes the shortest or full image dimensions predictable before cropping.
- **Crop:** selects the spatial region. Random crop augments training data; center crop keeps validation stable.
- **Flip:** adds left-right invariance when the label should not change under reflection.
- **Color jitter:** varies brightness, contrast, saturation, or hue so the model is less tied to lighting.
- **ToTensor:** converts a PIL/NumPy image from `[H, W, C]` values in `0..255` to `[C, H, W]` float values in `0..1`.
- **Normalize:** subtracts channel means and divides by channel standard deviations.

A good validation transform should be deterministic. If validation metrics move just because transforms are random, the experiment becomes hard to trust.


```python
import csv
import math
import random
from pathlib import Path

import numpy as np
from PIL import Image, ImageDraw, ImageEnhance

try:
    import torch
    TORCH_AVAILABLE = True
except ImportError:
    torch = None
    TORCH_AVAILABLE = False
    print("PyTorch is not installed. Tensor-based transform tests will be skipped.")

try:
    import torchvision.transforms as T
    TV_AVAILABLE = True
except Exception as exc:
    T = None
    TV_AVAILABLE = False
    print(f"torchvision is unavailable, using the local transform fallback: {type(exc).__name__}")

try:
    import matplotlib.pyplot as plt
    MPL_AVAILABLE = True
except Exception:
    plt = None
    MPL_AVAILABLE = False

SEED = 42
DATA_DIR = Path("_day11_image_data")
IMAGE_DIR = DATA_DIR / "images"
LABELS_CSV = DATA_DIR / "labels.csv"
MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)


def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    if TORCH_AVAILABLE:
        torch.manual_seed(seed)


set_seed(SEED)
print("torch:", TORCH_AVAILABLE, "torchvision:", TV_AVAILABLE)
```

## Prepared Image Data

Run this cell before the exercises. It writes a deterministic toy image classification dataset into `_day11_image_data/` so the transform work can focus on preprocessing and visualization instead of data collection.


```python
def make_transform_demo_dataset(out_dir=DATA_DIR, image_size=48, n_per_class=4, seed=42):
    rng = np.random.default_rng(seed)
    image_dir = out_dir / "images"
    image_dir.mkdir(parents=True, exist_ok=True)

    records = []
    classes = ["vertical", "horizontal", "diagonal"]
    base_colors = {
        "vertical": (220, 70, 80),
        "horizontal": (70, 170, 90),
        "diagonal": (65, 120, 220),
    }

    for label in classes:
        for idx in range(n_per_class):
            bg = np.zeros((image_size, image_size, 3), dtype=np.uint8)
            bg[:, :] = rng.integers(18, 55, size=3, dtype=np.uint8)
            img = Image.fromarray(bg, mode="RGB")
            draw = ImageDraw.Draw(img)
            color = tuple(min(255, max(0, c + int(rng.integers(-20, 21)))) for c in base_colors[label])
            width = 5 + idx % 3

            if label == "vertical":
                x = int(image_size * (0.25 + 0.14 * idx))
                draw.rectangle([x, 5, x + width, image_size - 6], fill=color)
                draw.ellipse([image_size - 16, 8, image_size - 6, 18], fill=(245, 235, 90))
            elif label == "horizontal":
                y = int(image_size * (0.25 + 0.14 * idx))
                draw.rectangle([5, y, image_size - 6, y + width], fill=color)
                draw.ellipse([8, image_size - 18, 18, image_size - 8], fill=(245, 235, 90))
            else:
                offset = idx - 1
                draw.line([4, image_size - 8 - offset, image_size - 6, 5 + offset], fill=color, width=width)
                draw.rectangle([8, 8, 18, 18], outline=(245, 235, 90), width=2)

            noise = rng.normal(0, 7, size=(image_size, image_size, 3))
            arr = np.clip(np.asarray(img, dtype=np.float32) + noise, 0, 255).astype(np.uint8)
            img = Image.fromarray(arr, mode="RGB")
            filename = f"{label}_{idx:02d}.png"
            rel_path = Path("images") / filename
            img.save(image_dir / filename)
            records.append({"image_path": str(rel_path).replace("\\", "/"), "label": label})

    with (out_dir / "labels.csv").open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["image_path", "label"])
        writer.writeheader()
        writer.writerows(records)

    return records


records = make_transform_demo_dataset()
print(f"Wrote {len(records)} images to {IMAGE_DIR}")
```

## Transform Helper Functions

These helpers make the notebook runnable even when `torchvision` is unavailable. When `torchvision` is installed, your transform builder may use `torchvision.transforms`; otherwise these helpers cover the same core ideas with PIL and PyTorch tensors.


```python
def load_rgb_image(path):
    return Image.open(path).convert("RGB")


def resize_pil(img, size):
    return img.resize((size, size), Image.BILINEAR)


def center_crop_pil(img, size):
    w, h = img.size
    left = max(0, (w - size) // 2)
    top = max(0, (h - size) // 2)
    return img.crop((left, top, left + size, top + size))


def random_crop_pil(img, size, rng):
    w, h = img.size
    if w < size or h < size:
        img = resize_pil(img, max(size, w, h))
        w, h = img.size
    left = rng.randint(0, w - size)
    top = rng.randint(0, h - size)
    return img.crop((left, top, left + size, top + size))


def random_horizontal_flip_pil(img, p, rng):
    if rng.random() < p:
        return img.transpose(Image.FLIP_LEFT_RIGHT)
    return img


def color_jitter_pil(img, brightness=0.25, contrast=0.25, rng=None):
    rng = rng or random.Random()
    b_factor = 1.0 + rng.uniform(-brightness, brightness)
    c_factor = 1.0 + rng.uniform(-contrast, contrast)
    img = ImageEnhance.Brightness(img).enhance(b_factor)
    img = ImageEnhance.Contrast(img).enhance(c_factor)
    return img


def pil_to_tensor(img):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required to convert images to tensors.")
    arr = np.asarray(img, dtype=np.float32) / 255.0
    return torch.from_numpy(arr).permute(2, 0, 1).contiguous()


def normalize_tensor(tensor, mean=MEAN, std=STD):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for normalization.")
    mean_t = torch.tensor(mean, dtype=tensor.dtype).view(3, 1, 1)
    std_t = torch.tensor(std, dtype=tensor.dtype).view(3, 1, 1)
    return (tensor - mean_t) / std_t


def denormalize_tensor(tensor, mean=MEAN, std=STD):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for denormalization.")
    mean_t = torch.tensor(mean, dtype=tensor.dtype).view(3, 1, 1)
    std_t = torch.tensor(std, dtype=tensor.dtype).view(3, 1, 1)
    return tensor * std_t + mean_t


def tensor_to_uint8_image(tensor):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for tensor conversion.")
    tensor = tensor.detach().cpu().clamp(0, 1)
    arr = (tensor.permute(1, 2, 0).numpy() * 255).round().astype(np.uint8)
    return Image.fromarray(arr, mode="RGB")


def apply_transform(transform, img, seed=None):
    if seed is not None:
        set_seed(seed)
        rng = random.Random(seed)
    else:
        rng = random.Random()
    try:
        return transform(img, rng=rng)
    except TypeError:
        return transform(img)


def stack_tensors(tensors):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for tensor batching.")
    return torch.stack(list(tensors), dim=0)
```

## Exercise 11-A: Build Train and Validation Transforms

Create two callable pipelines:

- `train_transforms`: resize, random crop, random horizontal flip, color jitter, tensor conversion, normalize.
- `val_transforms`: resize, center crop, tensor conversion, normalize.

Both should return tensors shaped `[3, image_size, image_size]`. The validation pipeline should be deterministic.


```python
def build_transform_pipelines(image_size=32, mean=MEAN, std=STD, use_torchvision=TV_AVAILABLE):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for tensor transforms.")

    resize_size = image_size + 8

    if use_torchvision and TV_AVAILABLE:
        train_pipeline = T.Compose([
            T.Resize((resize_size, resize_size)),
            T.RandomCrop(image_size),
            T.RandomHorizontalFlip(p=0.5),
            T.ColorJitter(brightness=0.25, contrast=0.25),
            T.ToTensor(),
            T.Normalize(mean=mean, std=std),
        ])
        val_pipeline = T.Compose([
            T.Resize((resize_size, resize_size)),
            T.CenterCrop(image_size),
            T.ToTensor(),
            T.Normalize(mean=mean, std=std),
        ])
        return train_pipeline, val_pipeline

    def train_pipeline(img, rng=None):
        rng = rng or random.Random()
        img = resize_pil(img, resize_size)
        img = random_crop_pil(img, image_size, rng)
        img = random_horizontal_flip_pil(img, p=0.5, rng=rng)
        img = color_jitter_pil(img, brightness=0.25, contrast=0.25, rng=rng)
        return normalize_tensor(pil_to_tensor(img), mean=mean, std=std)

    def val_pipeline(img, rng=None):
        img = resize_pil(img, resize_size)
        img = center_crop_pil(img, image_size)
        return normalize_tensor(pil_to_tensor(img), mean=mean, std=std)

    return train_pipeline, val_pipeline


train_transforms, val_transforms = build_transform_pipelines()
print("Built train_transforms and val_transforms")
```

## Exercise 11-B: Load Records and Transform a Batch

Read `labels.csv`, load images from disk, apply a transform, and return a mini-batch. This is the bridge between file-based datasets and model-ready tensors.


```python
def load_image_records(labels_csv=LABELS_CSV, data_dir=DATA_DIR):
    loaded = []
    with labels_csv.open("r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            image_path = data_dir / row["image_path"]
            loaded.append({"image_path": image_path, "label": row["label"]})
    return loaded


def transform_batch(records, transform, n=4, seed=42):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for batching transformed tensors.")
    tensors = []
    labels = []
    for i, record in enumerate(records[:n]):
        img = load_rgb_image(record["image_path"])
        tensors.append(apply_transform(transform, img, seed=seed + i))
        labels.append(record["label"])
    return stack_tensors(tensors), labels


loaded_records = load_image_records()
sample_batch, sample_labels = transform_batch(loaded_records, val_transforms, n=4)
print(sample_batch.shape, sample_labels)
```

## Exercise 11-C: Build the Augmentation Grid

Apply the training transform several times to the same source image and save a grid to `_day11_image_data/augmentation_grid.png`. The grid is the daily output/evidence for this lesson.


```python
def make_augmentation_grid(records, train_transform, index=0, n=16, cols=4, seed=42, out_path=DATA_DIR / "augmentation_grid.png"):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for the augmentation grid.")
    record = records[index]
    src = load_rgb_image(record["image_path"])
    rows = math.ceil(n / cols)
    tiles = []

    for i in range(n):
        transformed = apply_transform(train_transform, src.copy(), seed=seed + i)
        tile = tensor_to_uint8_image(denormalize_tensor(transformed))
        tiles.append(tile)

    tile_w, tile_h = tiles[0].size
    grid = Image.new("RGB", (cols * tile_w, rows * tile_h), color=(25, 25, 25))
    for i, tile in enumerate(tiles):
        x = (i % cols) * tile_w
        y = (i // cols) * tile_h
        grid.paste(tile, (x, y))

    out_path.parent.mkdir(parents=True, exist_ok=True)
    grid.save(out_path)
    return grid


augmentation_grid = make_augmentation_grid(loaded_records, train_transforms)
print("saved:", DATA_DIR / "augmentation_grid.png", "size:", augmentation_grid.size)
```

## Exercise 11-D: Compare Train vs Validation Behavior

Measure how much repeated outputs vary. Training transforms should vary because of random crop/flip/jitter. Validation transforms should stay fixed.


```python
def summarize_transform_behavior(image_path, train_transform, val_transform, repeats=6, seed=42):
    if not TORCH_AVAILABLE:
        raise ImportError("PyTorch is required for transform summaries.")
    img = load_rgb_image(image_path)
    train_outputs = [apply_transform(train_transform, img.copy(), seed=seed + i) for i in range(repeats)]
    val_outputs = [apply_transform(val_transform, img.copy(), seed=seed + i) for i in range(repeats)]

    train_stack = stack_tensors(train_outputs)
    val_stack = stack_tensors(val_outputs)
    train_variability = (train_stack - train_stack.mean(dim=0, keepdim=True)).abs().mean().item()
    val_variability = (val_stack - val_stack.mean(dim=0, keepdim=True)).abs().mean().item()

    return {
        "train_shape": tuple(train_stack.shape),
        "val_shape": tuple(val_stack.shape),
        "train_variability": train_variability,
        "val_variability": val_variability,
        "train_mean": train_stack.mean().item(),
        "val_mean": val_stack.mean().item(),
    }


behavior_summary = summarize_transform_behavior(loaded_records[0]["image_path"], train_transforms, val_transforms)
behavior_summary
```

## Test Cases

Run this cell after completing the TODO cells above. A correct implementation should print `Day 11 tests passed`.


```python
def run_day11_tests():
    if not TORCH_AVAILABLE:
        print("Day 11 tests skipped because PyTorch is not installed.")
        return

    required_names = [
        "make_transform_demo_dataset",
        "load_rgb_image",
        "build_transform_pipelines",
        "load_image_records",
        "transform_batch",
        "make_augmentation_grid",
        "summarize_transform_behavior",
    ]
    for name in required_names:
        assert name in globals(), f"Missing function: {name}"
        assert callable(globals()[name]), f"{name} must be callable"

    assert LABELS_CSV.exists(), "labels.csv was not created"
    assert IMAGE_DIR.exists(), "image directory was not created"
    image_files = sorted(IMAGE_DIR.glob("*.png"))
    assert len(image_files) == 12, f"Expected 12 demo images, found {len(image_files)}"

    rows = load_image_records()
    assert len(rows) == 12, f"Expected 12 records, found {len(rows)}"
    labels = sorted({row["label"] for row in rows})
    assert labels == ["diagonal", "horizontal", "vertical"], labels
    assert all(row["image_path"].exists() for row in rows), "Every record needs an existing image path"

    train_tf, val_tf = build_transform_pipelines(image_size=32)
    sample_img = load_rgb_image(rows[0]["image_path"])
    train_tensor = apply_transform(train_tf, sample_img.copy(), seed=123)
    val_tensor = apply_transform(val_tf, sample_img.copy(), seed=123)
    assert train_tensor.shape == (3, 32, 32), train_tensor.shape
    assert val_tensor.shape == (3, 32, 32), val_tensor.shape
    assert train_tensor.dtype == torch.float32
    assert val_tensor.dtype == torch.float32
    assert torch.isfinite(train_tensor).all()
    assert torch.isfinite(val_tensor).all()

    batch, batch_labels = transform_batch(rows, val_tf, n=5, seed=99)
    assert batch.shape == (5, 3, 32, 32), batch.shape
    assert len(batch_labels) == 5
    assert set(batch_labels).issubset({"vertical", "horizontal", "diagonal"})

    grid_path = DATA_DIR / "augmentation_grid.png"
    grid = make_augmentation_grid(rows, train_tf, index=0, n=16, cols=4, seed=7, out_path=grid_path)
    assert grid_path.exists(), "augmentation grid was not saved"
    assert grid.size == (128, 128), grid.size

    summary = summarize_transform_behavior(rows[0]["image_path"], train_tf, val_tf, repeats=5, seed=500)
    assert summary["train_shape"] == (5, 3, 32, 32), summary["train_shape"]
    assert summary["val_shape"] == (5, 3, 32, 32), summary["val_shape"]
    assert summary["train_variability"] > 0.001, summary
    assert summary["val_variability"] < 1e-6, summary

    print("Day 11 tests passed")


run_day11_tests()
```
